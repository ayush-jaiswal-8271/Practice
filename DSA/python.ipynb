```python
# take input of age and on the basis of age print whether the person is eligible or not eligible for drivimg licence
age = int(input("Enter your age: "))

if age >= 18:
    test = input("Enter the result of your driving test (PASS/FAIL): ")
    if test == "PASS":
        print("You are eligible for driving licence")
    else:
        print("You are not eligible for driving licence")
else:
    print("You are not eligible for driving licence")
```

```python
# answer[i] == "Fizzbuzz" if i % 3 == 0 and i % 5 == 0 else "Fizz" if i % 3 == 0 else "Buzz" if i % 5 == 0 else str(i)
ans = []
for i in range(1, int(input("Enter the value of n: "))+1):
    if i % 3 == 0 and i % 5 == 0:
        ans.append("FizzBuzz")
    elif i % 3 == 0:
        ans.append("Fizz")
    elif i % 5 == 0:
        ans.append("Buzz")
    else:
        ans.append(str(i))
print(ans)
```

```python
# Given two non-negative integers low and high. Return a list of all odd numbers in the inclusive range [low, high] in ascending order.
i = int(input("Enter the value of low: "))
j = int(input("Enter the value of high: "))

[i for i in range(i, j + 1) if i % 2 == 1]
```
