```python
# Given the array nums, for each nums[i] find out how many numbers in the array are smaller than it. That is, for each nums[i] you have to count the number of valid j's such that j != i and nums[j] < nums[i]. Return the answer in an array.
nums = list(map(int, input("Enter the elements of the array: ").split()))
ans  = []
for i in nums:
    c = 0
    for j in nums:
        if j < i:
            c += 1
    ans.append(c)
print(ans)
```

```python
# Given an integer num, return the number of digits in num that divide num. An integer val divides num if val is a divisor of num. Note: 0 does not divide any number.
num = int(input("Enter the number: "))
temp = num
ans = 0
while temp > 0:
    digit = temp % 10
    if digit != 0 and num % digit == 0:
        ans += 1
    temp //= 10
print(ans)
```

```python
# Given an integer x, return true if x is a palindrome, and false otherwise.
x = int(input("Enter the number: "))
temp = x
rev = 0 
while temp > 0:
    dig = temp % 10
    rev = rev * 10 + dig
    temp //= 10
if x == rev:
    print("True")
else:
    print("False")
```

```python
# Given an integer number n, return the difference between the product of its digits and the sum of its digits.
n = int(input("Enter the number: "))
temp = n
prod = 1
sum = 0
while temp > 0:
    dig = temp % 10
    prod *= dig
    sum += dig
    temp //= 10
print(prod - sum)
```

```python
#There are n kids with candies. You are given an integer array candies, where each candies[i] represents the number of candies the ith kid has, and an integer extraCandies, denoting the number of extra candies that you have. Return a boolean array result of length n, where result[i] is true if, after giving the ith kid all the extraCandies, they will have the greatest number of candies among all the kids, or false otherwise.
candies = list(map(int, input("Enter the number of candies each kid has: ").split()))
extraCandies = int(input("Enter the number of extra candies: "))
max_candies = max(candies)
ans = []
for i in candies:
    if i + extraCandies >= max_candies:
        ans.append(True)
    else:
        ans.append(False)
print(ans)
```
